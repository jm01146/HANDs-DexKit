```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

df_0 = pd.read_excel('distanceRecordingDataXAXIS.xlsx')
df_1 = pd.read_excel('distanceRecordingDataXAXIS_z3.xlsx')
df_2 = pd.read_excel('distanceRecordingDataXAXIS_z5.xlsx')

df_0["Time"] = pd.to_datetime(df_0["timestamp_iso"])
df_0["ElapsedTime"] = (df_0["Time"] - df_0["Time"].iloc[0]).dt.total_seconds()

df_1["Time"] = pd.to_datetime(df_1["timestamp_iso"])
df_1["ElapsedTime"] = (df_1["Time"] - df_1["Time"].iloc[0]).dt.total_seconds()

df_2["Time"] = pd.to_datetime(df_2["timestamp_iso"])
df_2["ElapsedTime"] = (df_2["Time"] - df_2["Time"].iloc[0]).dt.total_seconds()
```

# Studying delta uT within the X axis

```python
plt.plot(df_0["ElapsedTime"], df_0["Bx_uT"], label="Z Heigth: 0mm")
plt.xlabel("Time (seconds)")
plt.ylabel("Bx (uT)")
plt.title("Magnetic Flux Density @ x-axis vs Time")
plt.legend()
plt.show()
```

```python
plt.plot(df_1["ElapsedTime"], df_1["Bx_uT"], label="Z Heigth: 3mm")

plt.xlabel("Time (seconds)")
plt.ylabel("Bx (uT)")
plt.title("Magnetic Flux Density @ x-axis vs Time")
plt.legend()
plt.show()
```

```python
plt.plot(df_0["ElapsedTime"], df_0["Bx_uT"], label="Z Heigth: 0mm")
plt.plot(df_1["ElapsedTime"], df_1["Bx_uT"], label="Z Heigth: 3mm")
plt.plot(df_2["ElapsedTime"], df_2["Bx_uT"], label="Z Heigth: 5mm")

plt.xlabel("Time (seconds)")
plt.ylabel("Bx (uT)")
plt.title("Magnetic Flux Density @ x-axis vs Time")
plt.legend()
plt.show()
```
